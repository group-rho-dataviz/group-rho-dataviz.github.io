# Some Additional Info on Top News

```python
import pandas as pd
import numpy as np
```

```python
data_path = '../data/processed/weekly_top_25_material_conflict.csv'
df = pd.read_csv(data_path)
```

```python
df.head()
```

```python
# drop columns ['material_conflict_unique_events', 'material_conflict_mentions_weighted']
df = df.drop(columns=['material_conflict_unique_events', 'material_conflict_mentions_weighted'])
df.head()
```

```python
# For each 'mention_week' keep rows with highest 5 'material_conflict_mentions'
df_filtered = df.groupby('mention_week').apply(lambda x: x.nlargest(5, 'material_conflict_mentions')).reset_index(drop=True)
df_filtered.head()
```

```python
df_filtered.head(10)
```

```python
# Add 'google_search' column 
newspaper_sources = ['apnews.com', 'reuters.com', 'reuters.com', 'npr.org', 'pbs.org', 'bbc.com', 'dw.com', 'aljazeera.com']

# Ensure your date column is actually datetime objects (not strings)
df_filtered['mention_week'] = pd.to_datetime(df_filtered['mention_week'])

# This creates a list of random sources the same length as your DataFrame
random_choices = np.random.choice(newspaper_sources, size=len(df_filtered))

# We cast to string and take the date part only (YYYY-MM-DD)
start_date = df_filtered['mention_week'].dt.strftime('%Y-%m-%d')
end_date = (df_filtered['mention_week'] + pd.Timedelta(days=7)).dt.strftime('%Y-%m-%d')

df_filtered['google_search'] = (
    "https://www.google.com/search?q=site:" + 
    random_choices + 
    "+" + df_filtered['conflict_country_name'] + 
    "+news+after:" + start_date + 
    "+before:" + end_date
)

# Optional: Print one to verify
print(df_filtered['google_search'].iloc[0])
```
