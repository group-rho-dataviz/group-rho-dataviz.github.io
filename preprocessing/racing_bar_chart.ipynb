# Racing Bar Chart - Data Analysis Notebook

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import json
```

```python
# Read json files from ../src/json/fips.json
# FIPS 10-4 to Country Name mapping
with open('../src/json/fips.json', 'r') as f:
    fips = json.load(f)
```

```python
# Function to convert 2-letter FIPS 10-4 country codes to country names
def code_to_country(code):
    try:
        country = fips[code]
        return country
    except:
        raise ValueError(f"Invalid country code: {code}")
```

```python
# Load src/data/processed/racing_line_chart.csv as pandas dataframe
racing_data = pd.read_csv('../data/processed/racing_line_chart.csv')
racing_data.head()
```

```python
# Grouping by 'conflict_country_name' create a new column 'cum_material_conflict_mentions' 
# that is the cumulative sum of 'material_conflict_mentions' for that 'conflict_country_name'
racing_data['cum_material_conflict_mentions'] = racing_data.groupby(['conflict_country_name'])['material_conflict_mentions'].cumsum()
racing_data.head()
```

```python
# Showing only the rows where 'conflict_country_name' is 'Syria' orderd by 'mention_week'
racing_data[racing_data['conflict_country_name'] == 'Syria'].sort_values('mention_week')
```

```python
racing_data['cum_material_conflict_mentions_weighted'] = racing_data.groupby(['conflict_country_name'])['material_conflict_mentions_weighted'].cumsum()
racing_data['cum_material_conflict_unique_events'] = racing_data.groupby(['conflict_country_name'])['material_conflict_unique_events'].cumsum()
```

```python
racing_data.head()
```

```python
racing_data[racing_data['conflict_country_name'] == 'Syria'].sort_values('mention_week')
```

```python
# Drop the columns 'material_conflict_mentions', 'material_conflict_mentions_weighted', 'material_conflict_unique_events'
racing_data = racing_data.drop(columns=['material_conflict_mentions', 'material_conflict_mentions_weighted', 'material_conflict_unique_events'])
```

```python
# Rename the columns to 'material_conflict_mentions', 'material_conflict_mentions_weighted', 'material_conflict_unique_events' respectively
racing_data = racing_data.rename(columns={
    'cum_material_conflict_mentions': 'material_conflict_mentions',
    'cum_material_conflict_mentions_weighted': 'material_conflict_mentions_weighted',
    'cum_material_conflict_unique_events': 'material_conflict_unique_events'
})
```

```python
# Saving the processed dataframe to ../data/processed/racing_bar_chart.csv
racing_data.to_csv('../data/processed/racing_bar_chart.csv', index=False)
```
